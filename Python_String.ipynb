```python

#1
s = input("Enter a string :")
q = s[::-1]
if s==q:
    print("String is palindrome")
else :
    print("String is not palindrome")
```

```python
#2
s = input("Enter a string :")
p = s[:2]+s[-2:]
if len(p)<2:
  print()
else:
  print(p)
```

```python
#3
print("Hello, %s your age %d and salary %f"%("Ashish",92,50000))
print("Hello, {} your age {} and salary {}".format("Ashish",92,50000))
print("Hello, {0} your age {1} and salary {2}".format("Ashish",92,50000))
print("Hello, {r} your age {q} and salary {s}".format(r="Ashish",q=92,s=50000))
```
